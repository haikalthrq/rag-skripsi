# 07 — Tabel Ringkasan Metrik & Export

Notebook ini menghasilkan:
1. Tabel ringkasan metrik rata-rata per metode (Top-5) — siap di-paste ke laporan
2. Tabel metrik per top-k untuk semua metode
3. Export ke CSV di `results/final/figures/summary_metrics_top5.csv`

> **Prasyarat:** Jalankan `00_load_data.ipynb` terlebih dahulu.

## 1. Import & Load Cache

```python
import pickle
from pathlib import Path

import pandas as pd

PKL_DIR = Path("_cache")

with open(PKL_DIR / "data_eval.pkl", "rb") as f:
    cache = pickle.load(f)

all_df  = cache["all_df"]
top5    = cache["top5"]
FIG_DIR = cache["FIG_DIR"]

METHOD_ORDER = ["Element-Based", "MaxMin Semantic", "Recursive"]
print(f"Data loaded.")
```

## 2. Tabel Ringkasan Top-5

```python
summary_top5 = (
    top5
    .groupby("method")[["precision_at_k","recall_at_k","mrr","bleu","rouge_l_recall"]]
    .mean()
    .reindex(METHOD_ORDER)
    .rename(columns={
        "precision_at_k": "Precision@5",
        "recall_at_k":    "Recall@5",
        "mrr":            "MRR",
        "bleu":           "BLEU",
        "rouge_l_recall": "ROUGE-L",
    })
    .round(4)
)
summary_top5.index.name = "Metode"

print("=== Rata-rata Metrik per Metode (Top-5) ===")
display(
    summary_top5
    .style
    .highlight_max(color="#c8e6c9", axis=0)
    .highlight_min(color="#ffcccc", axis=0)
    .format("{:.4f}")
)
```

## 3. Tabel Lengkap per Top-k

```python
summary_all_k = (
    all_df
    .groupby(["top_k", "method"])[["precision_at_k","recall_at_k","mrr","bleu","rouge_l_recall"]]
    .mean()
    .round(4)
    .rename(columns={
        "precision_at_k": "Precision@k",
        "recall_at_k":    "Recall@k",
        "mrr":            "MRR",
        "bleu":           "BLEU",
        "rouge_l_recall": "ROUGE-L",
    })
)

print("=== Metrik per Top-k per Metode ===")
display(summary_all_k)
```

## 4. Tabel per Tipe Query (Tabel vs Narasi) — Top-5

```python
summary_by_type = (
    top5
    .groupby(["query_type","method"])[["precision_at_k","recall_at_k","mrr","bleu","rouge_l_recall"]]
    .mean()
    .round(4)
    .rename(columns={
        "precision_at_k": "Precision@5",
        "recall_at_k":    "Recall@5",
        "mrr":            "MRR",
        "bleu":           "BLEU",
        "rouge_l_recall": "ROUGE-L",
    })
)

print("=== Metrik per Tipe Query per Metode (Top-5) ===")
display(summary_by_type)
```

## 5. Export ke CSV

```python
# Ringkasan top-5
out1 = FIG_DIR / "summary_metrics_top5.csv"
summary_top5.to_csv(out1)
print(f"Tersimpan: {out1.resolve()}")

# Lengkap per top-k
out2 = FIG_DIR / "summary_metrics_all_k.csv"
summary_all_k.to_csv(out2)
print(f"Tersimpan: {out2.resolve()}")

# Per tipe query
out3 = FIG_DIR / "summary_metrics_by_type.csv"
summary_by_type.to_csv(out3)
print(f"Tersimpan: {out3.resolve()}")
```
